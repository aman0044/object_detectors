## Image operations

### Loading an Image
We use the **`image_ops.load_image`** function to load an image:

- **`img_path`**: Specifies the path to the image.
- **`image_ops.load_image`**: A utility function from the `core_assist` library that loads the image in BGR format (by default).

Returns
- image

This ensures the image is loaded correctly and is ready for further processing.


```python
from core_assist import image_ops , plot
img_path = "/home/azureuser/cloudfiles/code/Users/rohit.chandra/scripts/cv_utility/examples/test_image/gettyimages-1387151277-612x612.jpg"

img = image_ops.load_img(img_path=img_path)
```

### Loading the Image in Different Formats
We use the following functions to load the image in multiple formats for diverse processing needs, with the single argument **`img_path`**:

- **`load_bgr(img_path)`**: Loads the image in BGR format (commonly used in OpenCV).  
- **`load_rgb(img_path)`**: Loads the image in RGB format (preferred for visualization).  
- **`load_gray(img_path)`**: Loads the image in grayscale, useful for tasks requiring intensity-based analysis. 
 
Returns
- image


```python
bgr_image =  image_ops.load_bgr(img_path)
rgb_image = image_ops.load_rgb(img_path)
gray_image = image_ops.load_gray(img_path)
plot.image(BGR=bgr_image, RGB=rgb_image, Gray=gray_image)
```

### Loading Image from Binary Buffer
After reading the image into a binary buffer, we use the following functions to load the image in different formats, which only require the **buffer_data** as an argument:

- **`load_buffer_bgr(buffer_data)`**: Loads the image in BGR format.  
- **`load_buffer_rgb(buffer_data)`**: Loads the image in RGB format.  
- **`load_buffer_gray(buffer_data)`**: Loads the image in grayscale.

Returns
- image


```python
with open(img_path , "rb") as f:
    buffer = f.read()
bgr_image = image_ops.load_buffer_bgr(buffer_data=buffer)
rgb_image = image_ops.load_buffer_rgb(buffer_data=buffer)
gray_image = image_ops.load_buffer_gray(buffer_data=buffer)
plot.image(BGR=bgr_image, RGB=rgb_image, Gray=gray_image)
```

### Saving the Image to a File
The **`image_ops.save`** function is used to save an image to the specified file and format:

- **`image`**: The image to be saved (in this case, `bgr_image`).
- **`filename`**: The name of the file where the image will be saved (without the extension).
- **`format`**: The image format to save in (e.g., `.jpeg`). The default format is `.jpg`.

Valid formats: `.jpg`, `.jpeg`, `.png`, `.bmp`, `.tiff`

```python
image_ops.save(bgr_image , filename="examples/test_image/3" , format=".jpeg")
```

### Resizing the Image
The **`image_ops.resize`** function resizes an image while optionally preserving the aspect ratio. It supports resizing regular images as well as mask images with different interpolation techniques.

**Arguments**
- **`image`** (`numpy.ndarray`): The input image.
- **`size`** (`Optional[Union[int, Tuple[int, int]]]`):  
  - If an `int` is provided, the image is resized while keeping the aspect ratio.  
  - If a `tuple (width, height)` is provided, the image is resized to exact dimensions.  
  - If `None`, the original image is returned.  
- **`is_mask`** (`bool`, optional):  
  - If `True`, uses nearest-neighbor interpolation (`INTER_NEAREST`), suitable for masks.  
  - If `False`, uses area interpolation (`INTER_AREA`), better for reducing image size smoothly.  

**Returns**
- A **resized `numpy.ndarray`** representing the resized image.

```python
img = image_ops.load_rgb(img_path)
resized_img = image_ops.resize(image=img, size=(200,200))
print(f"image size before resizing : {img.shape} \nimage size after resizing : {resized_img.shape}")
plot.image(original_image = img,resized_image=resized_img)
```

- **`image`**: The image to be cropped (in this case, `img`).
- **`x`**: The starting x-coordinate for the crop.
- **`y`**: The starting y-coordinate for the crop.
- **`width`**: The width of the cropped region.
- **`height`**: The height of the cropped region.

Returns

- **`image`**: The image to be rotated (in this case, `img`).

Returns
- rotated image

### Converting the Image to Different Color Spaces
The **`image_ops.convert_color_space`** function is used to convert an image to a specified color space:

- **`image`**: The input image to be converted (in this case, `img`).
- **`color_space`**: The target color space for conversion. Supported values are:
  - `"GRAY"`: Converts the image to grayscale.
  - `"RGB"`: Converts the image to RGB color space.
  - `"HSV"`: Converts the image to HSV color space.

Returns
- converted color image

```python
img = image_ops.load_bgr(img_path)
hsv_image = image_ops.convert_color_space(img,color_space="HSV")
rgb_image = image_ops.convert_color_space(img ,color_space="RGB")
gray_image = image_ops.convert_color_space(img,color_space="GRAY")
plot.image(original_image = img,hsv_image=hsv_image , rgb_image = rgb_image , gray_image = gray_image)
```

### Blending Two Images
The **`image_ops.blend`** function is used to blend two images together using specified weights:

- **`image1`**: The first input image.
- **`image2`**: The second input image.
- **`alpha`**: The weight for the first image.
- **`beta`**: The weight for the second image.
- **`gamma`** (optional): A scalar added to each sum, default is 0.

Returns
- blended image

```python
blended_image = image_ops.blend(image_1 , image_2,alpha=0.5 , beta=0.5,gamma=0.4)
plot.image(image_1 = image_1 , image_2 = image_2 , blended_image = blended_image)
```

### Adding Padding to the Image
The **`image_ops.add_padding`** function is used to add padding to an image with various border types:

- **`image`**: The input image (in this case, `image_2`).
- **`constant_value`**: A constant value to add padding to all sides. If provided, it will be applied to the top, bottom, left, and right borders.
- **`top`, `bottom`, `left`, `right`** (optional): The specific number of pixels to add to each side of the image.
- **`border_type`**: The type of border to add. Supported types are:
  - `'constant'`: Adds padding with a constant color (default is black).
  - `'reflect'`: Reflects the image to create the padding.
  - `'replicate'`: Replicates the edge pixels to create the padding.
  - `'wrap'`: Wraps the image around to create the padding.
- **`color`** (optional): The color of the padding (for `'constant'` border type). Default is black.

Returns
- padded image

```python
padded_image = image_ops.add_padding(image_2 , constant_value= 100 , border_type="constant" , color=(0,0,0) )
plot.image(orginal_image = image_2 , padded_image = padded_image )
```

### Detecting Edges in the Image
The **`image_ops.detect_edges`** function is used to detect edges in an image using various methods:

- **`image`**: The input image (should be in grayscale for most methods).
- **`method`**: The edge detection method to use:
  - `"canny"`: Uses the Canny edge detection algorithm.
  - `"sobel"`: Uses the Sobel operator to detect edges.
  - `"laplacian"`: Uses the Laplacian operator to detect edges.
- **`threshold1`, `threshold2`**: Thresholds used in Canny edge detection (optional).
- **`ksize`**: The kernel size for Sobel and Laplacian filters (optional; must be odd and greater than 1).


Returns
- edge detected image 

```python
canny_edges = image_ops.detect_edges(image=image_2 , method="canny" , threshold1=100,threshold2=200)
sobel_edges = image_ops.detect_edges(image_2 , method="sobel", ksize=3)
lapacian_edges = image_ops.detect_edges(image_2 , method="laplacian" , ksize=3)

plot.image(original_img = image_2 , canny_edges = canny_edges , sobel_edges = sobel_edges , lapacian_edges = lapacian_edges)
```

### Extracting Contours from an Image
The **`image_ops.extract_contours`** function is used to extract contours from a binary or grayscale image:

- **`image`**: The input binary or grayscale image..
- **`min_area`**: The minimum area a contour must have to be considered. Contours with an area smaller than this value are ignored.
- **`retrieval_mode`**: The contour retrieval mode. Two options:
  - `'external'`: Retrieves only the extreme outer contours.
  - `'tree'`: Retrieves all contours and reconstructs the full hierarchy of nested contours.
- **`approximation_mode`**: The contour approximation method:
  - `'simple'`: Compresses horizontal, vertical, and diagonal segments to make the contour simpler.
  - `'none'`: No approximation, keeps all the contour points.

Returns
- countours 
- herarichy

```python
import cv2
test_img4 = image_ops.load_rgb("/home/azureuser/cloudfiles/code/Users/rohit.chandra/scripts/cv_utility/examples/test_image/Front_Bumper_(Fascia)_Carrier_Single_Motor_RGB.png")
contours  , herarichy = image_ops.extract_contours(image=image_ops.detect_edges(test_img4), min_area=0,retrieval_mode="tree" , approximation_mode="simple")
image_with_contours4 = cv2.drawContours(test_img4.copy(), contours, -1, (255, 0, 0), 3)
plot.image(orinal_image = test_img4,image_with_contours = image_with_contours4)
```

### Blending Foreground and Background Using a Mask
The **`image_ops.blend_with_mask`** function blends a foreground image onto a background image using a binary mask:

- **`background`**: The background image.
- **`foreground`**: The foreground image that will be blended onto the background.
- **`mask`**: A binary mask where the white regions represent the foreground, and black regions represent the background.
- **`alpha`**: The weight for the foreground image (must be between 0 and 1).
- **`beta`**: The weight for the background image (must be between 0 and 1).
- **`gamma`**: A scalar value added to each weighted sum (default is 0).

Returns
- masked blended image


```python

blend_using_mask = image_ops.blend_with_mask(background=image_2 , foreground=image_1,mask=mask ,alpha=0.9 , beta=1,gamma=0)
plot.image(background = image_2, foreground = image_1 ,blend_using_mask = blend_using_mask)
```

### Cropping with Padding Within Image Bounds

The **`crop_with_pad`** function crops a specified region from an image while ensuring that optional padding remains within the original image bounds.

#### Parameters:
- **`image`** (`np.ndarray`): The input image as a NumPy array (in BGR format).  
- **`left`** (`int`): The x-coordinate of the left edge of the crop rectangle.  
- **`top`** (`int`): The y-coordinate of the top edge of the crop rectangle.  
- **`right`** (`int`): The x-coordinate of the right edge of the crop rectangle.  
- **`bottom`** (`int`): The y-coordinate of the bottom edge of the crop rectangle.  
- **`pad`** (`float`): Padding percentage to add to the crop area (value between 0 and 1). Default is `0.0`.

#### Returns:
- **`np.ndarray`**: The cropped image with padding applied, restricted to the bounds of the original image.


```python
cropped_with_pad = image_ops.crop_with_pad(image=image_2 ,left=100 , right=250 , top=115 , bottom=275 , pad=0.1)
print(image_2.shape ,cropped_with_pad.shape)
plot.image(original = image_2 , crop_with_pad = cropped_with_pad)
```

### Applying Blur to a Region of Interest (ROI)

The **`ROI_blur`** function applies a blur effect to a specific region of interest (ROI) within an image.

#### Parameters:
- **`image`** (`np.ndarray`): The input image in the shape `(H, W, C)`.  
- **`roi`** (`tuple`): The region of interest specified as `(x, y, width, height)`, where:
  - `x` and `y` are the coordinates of the top-left corner of the ROI.  
  - `width` and `height` specify the size of the ROI.  
- **`blur`** (`int`): The kernel size for the blur effect. If an even number is provided, it will be adjusted to the nearest odd number.

#### Returns:
- **`np.ndarray`**: The image with the specified ROI blurred, while the rest of the image remains unchanged.


```python
img_with_roi_blur = image_ops.ROI_blur(image_2,roi=(100,115,150,150) , blur=50)
print(image_2.shape)
plot.image(original = image_2 , img_with_roi_blur = img_with_roi_blur)
```
